```python
import pandas as pd
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/master/Salaries.csv')
```

```python
print(df.head())
```

```python
df = df.dropna()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.boxplot(data=df, x='rank', y='salary')
plt.title('Salary by Rank')
plt.show()
```

```python
sns.boxplot(data=df, x='rank', y='salary', hue='sex')
plt.title('Salary by Rank and Sex')
plt.legend(title='Sex', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
